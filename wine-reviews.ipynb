### DataFrame

`apply()`

```
Apply a function along an axis of the DataFrame
```

- https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.apply.html#pandas.DataFrame.apply

`map()`

```
Apply a function to a Dataframe elementwise
```

- https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.map.html#pandas.DataFrame.map


````python
review_points_mean = reviews.points.mean()
reviews.points.map(lambda p: p - review_points_mean)
````

````python
def remean_points(row):
    row.points = row.points - review_points_mean
    return row

reviews.apply(remean_points, axis='columns')
````

````python
reviews.country + " - " + reviews.province
````

````python
# which wine is the "best bargain"?
# => the wine with the highest points-to-price ratio

bargain_idx = (reviews.points / reviews.price).idxmax()
bargain_wine = reviews.loc[bargain_idx, 'title']

bargain_wine, bargain_idx
````

````python
# tropical or fruity?

n_tropical = reviews.description.map(lambda desc: "tropical" in desc).sum()
n_fruity = reviews.description.map(lambda desc: "fruity" in desc).sum()
descriptor_counts = pd.Series([n_tropical, n_fruity], index=['tropical', 'fruity'], name='counts')

descriptor_counts
````
