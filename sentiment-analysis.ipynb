```python
import pandas as pd
books = pd.read_csv('books_with_categories.csv')
```

```python
from transformers import pipeline
classifier = pipeline("text-classification",
                      model="j-hartmann/emotion-english-distilroberta-base",
                      top_k = None,
                      device = "mps")
classifier("I am so excited!")
```

```python
# example : anger
anger = "I'm absolutely irate. Don't you dare speak to me like that again."
classifier(anger)
```

```python
# example : disgust
disgust = "Ew! That food is spoiled!"
classifier(disgust)
```

```python
# example : fear
fear = "Her heart hammered against her ribs, a frantic drumbeat echoing the terror coiling in her stomach. Her breath hitched, shallow gasps catching in her throat, while a cold sweat broke out on her skin. She wanted to run, but her legs felt like lead, rooted to the spot in a paralysis of pure panic"
classifier(fear)
```

```python
# example : happiness
happiness = "I'm so excited for you."
classifier(happiness)
```

```python
# example : sadness
sadness = "I'm feeling blue."
classifier(sadness)
```

```python
# example : surprise
surprise = "Oh my goodness, that's unbelievable!"
classifier(surprise)
```

```python
import numpy as np

emotion_labels = ["anger", "disgust", "fear", "joy", "sadness", "surprise", "neutral"]
isbn = []
emotion_scores = {label: [] for label in emotion_labels}

def calculate_max_emotion_scores(predictions):
    per_emotion_scores = {label: [] for label in emotion_labels}
    for prediction in predictions:
        sorted_predictions = sorted(prediction, key=lambda x: x["label"])
        for index, label in enumerate(emotion_labels):
            per_emotion_scores[label].append(sorted_predictions[index]["score"])
    return {label: np.max(scores) for label, scores in per_emotion_scores.items()}
```

```python
for i in range(10):
    isbn.append(books["isbn13"][i])
    sentences = books["description"][i].split(".")
    predictions = classifier(sentences)
    max_scores = calculate_max_emotion_scores(predictions)
    for label in emotion_labels:
        emotion_scores[label].append(max_scores[label])
```

```python
from tqdm import tqdm

emotion_labels = ["anger", "disgust", "fear", "joy", "sadness", "surprise", "neutral"]
isbn = []
emotion_scores = {label: [] for label in emotion_labels}

for i in tqdm(range(len(books))):
    isbn.append(books["isbn13"][i])
    sentences = books["description"][i].split(".")
    predictions = classifier(sentences)
    max_scores = calculate_max_emotion_scores(predictions)
    for label in emotion_labels:
        emotion_scores[label].append(max_scores[label])
```

```python
emotions_df = pd.DataFrame(emotion_scores)
emotions_df["isbn13"] = isbn
```

```python
emotions_df
```

```python
books = pd.merge(books, emotions_df, on = "isbn13")
```
