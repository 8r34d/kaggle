## Examples of Degrees of Freedom

### Example 1 : n - 1
- Consider a data sample consisting of five positive integers
- The values of the five integers must have an average of six
- If four items within the data set are {3, 8, 5, and 4}, the fifth number must be 10
- Because the first four numbers can be chosen at random, the degree of freedom is four

### Example 2 : n OR n - 0
- Consider a data sample consisting of five positive integers
- The values could be any number with no known relationship between them
- In other words, there are no constraints and limitations on the numbers selected
- Because all five numbers can be selected randomly and with no limitations, the degree of freedom is five

```python
example_one = pd.Series([3,4,5,8,10])
example_one.mean()
```

```python
ra = np.sum(example_one[:]) / len(example_one)
ra == example_one.mean()
```

```python
first_four = example_one[:-1]
first_four
```

```python

rb = np.sum(first_four) / len(example_one)
```

```python
missing_value = (ra - rb) * len(example_one)
missing_value == 10
```

## Population Standard Deviation
- When calculating the standard deviation for an entire population, there are no restrictions on the data points, so the divisor is N (the total number of elements)
- In this case, ddof = 0, so the divisor is N - 0 = N.
- Default for NumPy

```python
# pandas >> population standard deviation : n
# ddof : int, default 1
# Delta Degrees of Freedom
# The divisor used in calculations is N - ddof, where N represents the number of elements
s.std(ddof=0)
```

```python
# numpy >> population standard deviation : n
# ddof : int, float, default 0
np.std(s)
```

## Sample Standard Deviation
- When calculating standard deviation for a sample, the sample mean is usually used as an intermediate parameter
- This is a restriction because the last data point is no longer free to vary once the mean is set
- The sample size is reduced by one to account for this restriction, making the divisor N - 1
- In this case, ddof = 1
- Default for pandas

```python
# pandas >> sample standard deviation : n - 1
# ddof : int, default 1
# Delta Degrees of Freedom
# The divisor used in calculations is N - ddof, where N represents the number of elements
s.std()
```

```python
# numpy >> sample standard deviation : n - 1
# ddof : int, float, default 0
np.std(s, ddof=1)
```

```python
a = s[0] - s.mean()
b = s[1] - s.mean()
c = s[2] - s.mean()
d = s[3] - s.mean()
e = s[4] - s.mean()

total = (a ** 2) + (b ** 2) + (c ** 2) + (d ** 2) + (e ** 2)
```

```python
# population standard deviation : n << numpy
population = total / len(s)
psd = np.sqrt(population)
psd
```

```python
# sample standard deviation : n - 1 << pandas
sample = total / (len(s) - 1)
ssd = np.sqrt(sample)
ssd
```
