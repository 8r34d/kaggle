# Series Examples

- https://pandas.pydata.org/docs/reference/api/pandas.Series.html#pandas.Series

```python
import pandas as pd
import numpy as np
```

```python
d1 = {'a': 1, 'b': 2, 'c': 3}
s1 = pd.Series(data=d1, index=['a', 'b', 'c'])
s1

# the keys of the dictionary match with the index values
# hence the index values have no effect
```

```python
d1.keys(), s1.index
```

```python
d2 = {'a': 1, 'b': 2, 'c': 3}
s2 = pd.Series(data=d2, index=['x', 'y', 'z'])
s2

# note that the index is first built with the keys from the dictionary
# after this the series is re-indexed with the given index values
# hence we get all NaN as a result
```

```python
d2.keys(), s2.index
```

```python
r1 = [1, 2]
s3 = pd.Series(r1, copy=False)
s3.iloc[0] = 999
s3

# constructing Series from a list with copy=False
```

```python
r1

# input data type : list
# due to input data type the Series has a `copy` of the original data
# even though copy=False => data is unchanged!
```

```python
r2 = [3, 4]
s4 = pd.Series(r1, copy=True)
s4.iloc[0] = 888
s4

# constructing Series from a list with copy=True
```

```python
r2

# copy=True => data is unchanged
```

```python
r3 = np.array([5, 6])
s5 = pd.Series(r3, copy=False)
s5.iloc[0] = 777
s5

# constructing Series from a 1d ndarray with copy=False
```

```python
r3

# input data type : 1d ndarray
# due to input data type the Series has a `view` on the original data
# copy=False => data is changed
```

```python
r4 = np.array([7, 8])
s6 = pd.Series(r4, copy=True)
s6.iloc[0] = 666
s6

# constructing Series from a 1d ndarray with copy=True
```

```python
r4

# copy=True => data is unchanged
```
