## Grouping and Sorting

### Series

`agg()`
```
Aggregate using one or more operations over the specified axis
```
- https://pandas.pydata.org/docs/reference/api/pandas.core.groupby.SeriesGroupBy.agg.html#pandas.core.groupby.SeriesGroupBy.agg

`groupby()`
```
Group Series using a mapper or by a Series of columns
```
- https://pandas.pydata.org/docs/reference/api/pandas.Series.groupby.html#pandas.Series.groupby

`reset_index()`
```
Generate a new DataFrame or Series with the index reset
This is useful when the index needs to be treated as a column
Or when the index is meaningless and needs to be reset to the default before another operation
```
- https://pandas.pydata.org/docs/reference/api/pandas.Series.reset_index.html#pandas.Series.reset_index

`sort_index()`
```
Sort Series by index labels
```
- https://pandas.pydata.org/docs/reference/api/pandas.Series.sort_index.html#pandas.Series.sort_index

`sort_values()`
```
Sort by the values
Sort a Series in ascending or descending order by some criterion
```
- https://pandas.pydata.org/docs/reference/api/pandas.Series.sort_values.html#pandas.Series.sort_values

### DataFrame

`agg()`
```
Aggregate using one or more operations over the specified axis
```
- https://pandas.pydata.org/docs/reference/api/pandas.core.groupby.DataFrameGroupBy.agg.html#pandas.core.groupby.DataFrameGroupBy.agg

`groupby()`
```
Group DataFrame using a mapper or by a Series of columns
```
- https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.groupby.html#pandas.DataFrame.groupby

`reset_index()`
```
Reset the index, or a level of it
Reset the index of the DataFrame, and use the default one instead
If the DataFrame has a MultiIndex, this method can remove one or more levels
```
- https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.reset_index.html#pandas.DataFrame.reset_index

`sort_index()`
```
Sort object by labels (along an axis)
```
- https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.sort_index.html#pandas.DataFrame.sort_index

`sort_values()`
```
Sort by the values along either axis
```
- https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.sort_values.html#pandas.DataFrame.sort_values

````python
reviews.groupby('points').points.count()
````

````python
reviews.groupby('points').price.min()
````

````python
reviews.groupby('winery').apply(lambda df: df.title.iloc[0], include_groups=False)
````

````python
reviews.groupby(['country', 'province']).apply(lambda df: df.loc[df.points.idxmax()], include_groups=False)
````

````python
reviews.groupby(['country']).price.agg([len, 'min', 'max'])
````

````python
countries_reviewed = reviews.groupby(['country', 'province']).description.agg([len])
countries_reviewed
````

````python
mi = countries_reviewed.index
type(mi)
````

````python
countries_reviewed.reset_index()

# convert multi-index back to regular index
````

````python
countries_reviewed = countries_reviewed.reset_index()
countries_reviewed.sort_values(by='len')
````

````python
countries_reviewed.sort_values(by='len', ascending=False)
````

````python
countries_reviewed.sort_values(by='province')
````

````python
countries_reviewed.sort_values(by='province', ascending=False)
````

````python
countries_reviewed.sort_index()
````

````python
countries_reviewed.sort_index(ascending=False)
````

````python
countries_reviewed.sort_values(by=['country', 'len'])
````

````python
countries_reviewed.sort_values(by=['country', 'len'], ascending=False)
````

````python
countries_reviewed.sort_values(by=['len', 'country'])
````

````python
countries_reviewed.sort_values(by=['len', 'country'], ascending=False)
````
