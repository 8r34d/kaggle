### DecisionTreeRegressor


`Decision Trees (DTs)`
- are a `non-parametric supervised learning` method used for classification and regression
- the goal is to create a model that predicts the `value` of a target variable by learning simple decision rules inferred from the data `features`
- a tree can be seen as a `piecewise constant approximation`

`Non parametric`
- refers to statistical methods and tests that make minimal or no assumptions about the shape of the data's underlying population distribution

`Supervised learning`
- is a category of machine learning that uses labeled datasets to train algorithms to predict outcomes and recognize patterns

`Piecewise constant approximation`
- is a method to represent a continuous function or signal by a series of distinct, constant-value steps or rectangles, typically over a partition of the domain
- this approach simplifies complex data by breaking it into smaller, manageable segments, with each segment having a single, constant value

## Model Validation

### Evaluate

- determine how accurate the model's predictions are

```python
print("First in-sample predictions:", iowa_model.predict(X.head()))
print("Actual target values for those homes:", y.head().tolist())
```

### The Problem with "In-Sample" Scores

- the measure we just computed can be called an "in-sample" score
- we used a single "sample" of houses for both building the model and evaluating it
- models' practical value come from making predictions on new data
- we measure performance on data that wasn't used to build the model

### Validation Data

- exclude some data from the model-building process
- test the model's accuracy on data it has not seen before

### Train Test Split
- `train_X, val_X, train_y, val_y = train_test_split(X, y, random_state = 0)`

```python
from sklearn.model_selection import train_test_split
```

```python
train_X, val_X, train_y, val_y = train_test_split(X, y, random_state=1)
```

```python
iowa_model_split = DecisionTreeRegressor(random_state=1)
```

```python
type(iowa_model_split)
```

```python
iowa_model_split.fit(train_X, train_y)
```

```python
val_predictions = iowa_model_split.predict(val_X)
```

```python
print("Validation predictions:", val_predictions[:5])
print("Validation prices:", val_y.head().tolist())
```

### Mean Absolute Error

- `val_mae = mean_absolute_error(val_y, val_predictions)`

```python
from sklearn.metrics import mean_absolute_error
val_mae = mean_absolute_error(val_y, val_predictions)
```

```python
print("Validation MAE: {:,.0f}".format(val_mae))
```

```python
def get_mae(max_leaf_nodes, train_features_x, val_features_x, train_target_y, val_target_y):
    model = DecisionTreeRegressor(max_leaf_nodes=max_leaf_nodes, random_state=0)
    model.fit(train_features_x, train_target_y)
    predictions_val = model.predict(val_features_x)
    mae = mean_absolute_error(val_target_y, predictions_val)
    return mae

print("get_mae complete")
```

```python
candidate_max_leaf_nodes = [5, 25, 50, 100, 250, 500]
scores = {leaf_size: get_mae(leaf_size, train_X, val_X, train_y, val_y) for leaf_size in candidate_max_leaf_nodes}
print(scores)
```

```python
best_tree_size = min(scores, key=scores.get)
print("best_tree_size: {:,.0f}\nmax_leaf_nodes: {:,.0f}".format(best_tree_size, scores[best_tree_size]))
```

```python
candidate_max_leaf_nodes_plus = [n for n in range(50, 100)]
scores_plus = {leaf_size: get_mae(leaf_size, train_X, val_X, train_y, val_y) for leaf_size in candidate_max_leaf_nodes_plus}
print(scores_plus)
```

```python
best_tree_size_plus = min(scores_plus, key=scores_plus.get)
print("best_tree_size_plus: {:,.0f}\nmax_leaf_nodes_plus: {:,.0f}".format(best_tree_size_plus, scores_plus[best_tree_size_plus]))
```

```python
final_model = DecisionTreeRegressor(max_leaf_nodes=best_tree_size_plus, random_state=1)
final_model.fit(X, y)
```

```python
final_predictions = final_model.predict(X)
final_mae = mean_absolute_error(y, final_predictions)
final_mae
```
